# Data Acquisition

Pulls the four datasets for the Chicago Traffic Crash Analytics project and
loads each into a dataframe. All datasets are on Socrata-based open data
portals, queried over plain HTTPS (no special SDK).

- Primary 1: Traffic Crashes - Crashes
- Primary 2: Traffic Crashes - People
- Secondary 1: NOAA hourly weather (Midway station)
- Secondary 2: Chicago Traffic Tracker - Historical Congestion Estimates by Segment


```python
import requests
import pandas as pd

pd.set_option("display.max_columns", 50)
```

## Primary 1: Traffic Crashes - Crashes

```python
CRASHES_URL = "https://data.cityofchicago.org/resource/85ca-t3if.json"

# Scope to a recent 2-year window to keep the pull manageable; full history
# is available from the same endpoint by removing/adjusting $where.
params = {
    "$limit": 50000,
    "$where": "crash_date >= '2023-01-01T00:00:00' AND crash_date < '2025-01-01T00:00:00'",
}

resp = requests.get(CRASHES_URL, params=params)
resp.raise_for_status()
crashes = pd.DataFrame(resp.json())
crashes.shape
```

```python
crashes.dtypes
```

```python
crashes.head()
```

## Primary 2: Traffic Crashes - People

```python
PEOPLE_URL = "https://data.cityofchicago.org/resource/u6pd-qa9d.json"

params = {
    "$limit": 50000,
    "$where": "crash_date >= '2023-01-01T00:00:00' AND crash_date < '2025-01-01T00:00:00'",
}

resp = requests.get(PEOPLE_URL, params=params)
resp.raise_for_status()
people = pd.DataFrame(resp.json())
people.shape
```

```python
people.dtypes
```

```python
people.head()
```

## Secondary 1: NOAA hourly weather (Midway station)

Requires a free NOAA CDO API token: https://www.ncdc.noaa.gov/cdo-web/token
Set it as an environment variable `NOAA_TOKEN` before running this cell.


```python
import os

NOAA_TOKEN = os.environ.get("NOAA_TOKEN")
NOAA_URL = "https://www.ncei.noaa.gov/cdo-web/api/v2/data"

# GHCND station for Midway Airport, Chicago: GHCND:USW00014819
params = {
    "datasetid": "GHCND",
    "stationid": "GHCND:USW00014819",
    "startdate": "2023-01-01",
    "enddate": "2023-01-31",
    "limit": 1000,
    "units": "standard",
}

if NOAA_TOKEN:
    resp = requests.get(NOAA_URL, headers={"token": NOAA_TOKEN}, params=params)
    resp.raise_for_status()
    weather = pd.DataFrame(resp.json().get("results", []))
else:
    weather = pd.DataFrame()
    print("NOAA_TOKEN not set, skipping live pull. See markdown above for how to get one.")

weather.shape
```

## Secondary 2: Chicago Traffic Tracker - Historical Congestion Estimates by Segment

```python
CONGESTION_URL = "https://data.cityofchicago.org/resource/4g9f-3jbs.json"

params = {
    "$limit": 50000,
    "$where": "time >= '2024-06-11T00:00:00' AND time < '2024-07-11T00:00:00'",
}

resp = requests.get(CONGESTION_URL, params=params)
resp.raise_for_status()
congestion = pd.DataFrame(resp.json())
congestion.shape
```
